```python
import os
import joblib
from xgboost import XGBRegressor
from sklearn.pipeline import Pipeline

# Chemin vers votre fichier
model_path = 'saved_models/xgbregressor_normalized_ENB2012_data.csv_l_heating_load.joblib'

# 1. Découpage du nom de fichier
basename = os.path.basename(model_path)                  # 'xgbregressor_normalized_ENB2012_data.csv_l_heating_load.joblib'
name_no_ext = basename[:-len('.joblib')]                 # 'xgbregressor_normalized_ENB2012_data.csv_l_heating_load'
class_part, rest = name_no_ext.split('_', 1)             # 'xgbregressor', 'normalized_ENB2012_data.csv_l_heating_load'
orig_csv, label = rest.split('.csv_', 1)                 # 'normalized_ENB2012_data.csv', 'l_heating_load'

# 2. Mapping nom de classe → classe Python
model_classes = {
    'pipeline': Pipeline,
    'xgbregressor': XGBRegressor,
    # ajouter d'autres mappings si besoin...
}

cls = model_classes.get(class_part.lower())
if cls is None:
    raise ValueError(f"Classe de modèle '{class_part}' non reconnue.")

# 3. Chargement du modèle
loaded_model = joblib.load(model_path)

# 4. Vérification du type
if not isinstance(loaded_model, cls):
    raise TypeError(f"Modèle chargé de type {type(loaded_model)} inattendu, "
                    f"attendu une instance de {cls}.")

# 5. Injection dans le namespace global sous le nom model_<label>
var_name = f"model_{label}"
globals()[var_name] = loaded_model

# Pour contrôler :
print(f"Modèle chargé dans la variable '{var_name}' ({cls.__name__}).")
```
